```python
import numpy as np
w = np.array([-2, 1, 0])
x = np.array([2, 3, 1])
y = 1
eta = 1.0
# Kiểm tra mẫu bị phân lớp sai
w_dot_x = np.dot(w, x)
is_misclassified = (y * w_dot_x <= 0)
print(f"Ban đầu w^T * x = {w_dot_x}")
print(f"Mẫu bị phân lớp sai không? {'Có' if is_misclassified else 'Không'}")
# Cập nhật Perceptron nếu sai
if is_misclassified:
    w_new = w + eta * y * x
    print(f"Trọng số w mới: {w_new}")
    # Tính lại w^T * x
    w_dot_x_new = np.dot(w_new, x)
    print(f"Sau cập nhật w^T * x = {w_dot_x_new}")
```
