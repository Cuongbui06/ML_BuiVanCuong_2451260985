```python
import numpy as np

class Perceptron:
    def __init__(self, toc_do_hoc=0.01, so_vong_lap=100):
        self.toc_do_hoc = toc_do_hoc
        self.so_vong_lap = so_vong_lap

    def fit(self, tap_x, tap_y):
        tap_x_mo_rong = np.hstack([tap_x, np.ones((len(tap_x), 1))])
        self.trong_so = np.zeros(tap_x_mo_rong.shape[1])

        # Huấn luyện
        for _ in range(self.so_vong_lap):
            for x_i, y_i in zip(tap_x_mo_rong, tap_y):
                if y_i * np.dot(x_i, self.trong_so) <= 0:
                    self.trong_so += self.toc_do_hoc * y_i * x_i

    def predict(self, tap_x):
        tap_x_mo_rong = np.hstack([tap_x, np.ones((len(tap_x), 1))])
        return np.where(np.dot(tap_x_mo_rong, self.trong_so) >= 0, 1, -1)

# Chạy thử
if __name__ == "__main__":
    x_phau_mau = np.array([[1, 2], [2, 3], [3, 1], [6, 5], [7, 8], [8, 6]])
    y_phau_mau = np.array([-1, -1, -1, 1, 1, 1])

    mo_hinh = Perceptron(toc_do_hoc=0.1, so_vong_lap=100)
    mo_hinh.fit(x_phau_mau, y_phau_mau)

    print("Nhãn dự đoán:", mo_hinh.predict(x_phau_mau))
```
